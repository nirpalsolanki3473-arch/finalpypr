```python
import pandas as pd
```

```python
df1=pd.read_csv('train.csv')
df1
```

```python
df1.drop_duplicates(inplace=True)
```

```python
df1.isnull().sum()
```

```python
df1['Age']=df1['Age'].fillna(df1['Age'].mean())
df1['Cabin']=df1['Cabin'].fillna('unknown')
df1['Embarked']=df1['Embarked'].fillna('unknown')
```

```python
df2=pd.read_csv('test.csv')
df2
```

```python
df3=pd.read_csv('gender_submission.csv')
df3
```

```python
fd=df2.join(df3,lsuffix='left',rsuffix='right')
fd
```

```python
fd.info()
```

```python
df1.info()
```

```python
fd.drop_duplicates(inplace=True)
```

```python
fd.drop('PassengerIdright',axis=1,inplace=True)
fd.rename(columns={'PassengerIdleft' : 'PassengerId' })
```

```python
fd.isnull().sum()
```

```python
fd['Fare']=fd['Fare'].fillna(fd['Fare'].mean())
fd['Age']=fd['Age'].fillna(fd['Age'].mean())
fd['Cabin']=fd['Cabin'].fillna('unknown')
```
